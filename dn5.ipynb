### 9. vprašanje

Z uporabo simbolov iz prejšnje naloge zapišite simbolni izraz za kinetično energijo rotorja. Kinetična energija rotacije je $E_k = \frac{1}{2}J\omega^2$, kjer je vztrajnostni moment rotorja $J = \frac{1}{2}m R^2$ in kotna hitrost $\omega = 2\pi f$. Izraz za $E_k$ poenostavite z uporabo funkcije ``sp.simplify()`` in dobljeni poenostavljeni simbolni izraz pretvorite v niz. (Namig: Za ulomke uporabite ``sp.Rational()`` in za pretvorbo v niz uporabite funkcijo ``str()``.)

Število poskusov: 3, enote:  /

```python
J_s = sp.Rational(1, 2) * m_s * R_s**2
omega_s = 2 * sp.pi * f_s
energija_s = sp.simplify(sp.Rational(1, 2) * J_s * omega_s**2)

odgovor9 = str(energija_s)
```

```python
poslji(odgovor9, id=5173325486326, st=9)
```

### 10. vprašanje

Izračunajte frekvenco vrtenja rotorja na sliki, če ste izmerili njegovo kinetično energijo $E_k$, polmer $R$ in maso $m$ (iz podatkov naloge). Rešitev podajte v obliki števila s plavajočo vejico na $n$ mest natančno. (Namig: Nastavite enačbo za $E_k$ in jo rešite za frekvenco $f$ ter zamenjajte simbolne spremenljivke z vrednostmi iz podatkov naloge.)

Število poskusov: 3, enote:  Hz

```python
enacba = sp.Eq(E_k_s, energija_s)
f_resitev = sp.solve(enacba, f_s)[0]

odgovor10 = f_resitev.subs({
    E_k_s: E_k,
    R_s: R,
    m_s: m
}).evalf(n)
```

### 12. vprašanje

Zaradi remonta morate rotor ustaviti. Ustavljate ga z nekonstantnim pojemkom $\sqrt{\frac{1}{2}t}$. Kotni zasuk rotorja je podan z enačbo $\theta(t) = 2 \pi \, f \, t - \frac{1}{2}\sqrt{\frac{1}{2}t} \, t^2$. Z uporabo simbolnega odvajanja dvakrat po času izračunajte kotni pospešek $\ddot{\theta}(t)$. Kot rezultat podajte vrednost kotnega pospeška ob času $t_0$ v obliki števila s plavajočo vejico na $n$ mest natančno (podatki iz naloge). (Namig: Za ulomke uporabite ``sp.Rational()``.)

Število poskusov: 3, enote:  1/$\text{s}^2$

```python
theta_s = (
    2 * sp.pi * f_s * t_s
    - sp.Rational(1, 2) * sp.sqrt(sp.Rational(1, 2) * t_s) * t_s**2
)

kotni_pospesek_s = sp.diff(theta_s, t_s, 2)
odgovor12 = kotni_pospesek_s.subs(t_s, t_0).evalf(n)
```

### 15. vprašanje

Stari rotor bi radi zamenjali z novim, ki ima spremenljivo površinsko gostoto $\rho(r) = -\ln{\left(\frac{r}{R}\right)}$, kjer je $R$ polmer (podatki naloge). Izračunajte celotno maso novega rotorja z dvojnim integralom v polarnih koordinatah: $m_{\text{novi}} = \int_0^{2\pi}\int_0^R \rho(r) \cdot r \, dr \, d\varphi$. Kot odgovor podajte maso novega rotorja kot število s plavajočo vejico na $n$ mest natančno.

Število poskusov: 3, enote:  kg

```python
r_s, phi_s = sp.symbols('r phi', real=True, positive=True)
rho_s = -sp.log(r_s / R_s)

masa_novega = sp.integrate(
    rho_s * r_s,
    (r_s, 0, R_s),
    (phi_s, 0, 2 * sp.pi)
)

odgovor15 = masa_novega.subs(R_s, R).evalf(n)
```

### 17. vprašanje

Rotor zavira zaradi zračnega trenja z momentom: $M_{u}=\frac{1}{2}(\dot{\theta}(t))^2$. Nastavite diferencialno enačbo za zasuk $\theta(t)$: $J\,\ddot{\theta}(t)+M_{u}=0$. Za masni vztrajnostni moment uporabite simbolni izraz od prej: $J=\frac{1}{2}m\,R^2$. (Namig: $\theta(t)$ definirajte kot splošno funkcijo z ``sp.Function('theta')(t)``, diferencialno enačbo nastavite z ``sp.Eq(...)``), za ulomke uporabite ``sp.Rational()``. Kot odgovor podajte poenostavljeno simbolno diferencialno enačbo (brez vstavljenih številskih vrednosti) v obliki niza.

Število poskusov: 3, enote:  /

```python
theta = sp.Function('theta')(t_s)

M_u = sp.Rational(1, 2) * sp.diff(theta, t_s)**2
enacba_gibanja = sp.Eq(J_s * sp.diff(theta, t_s, 2) + M_u, 0)

odgovor17 = str(sp.simplify(enacba_gibanja))
```

### 18. vprašanje

Zgornjo diferencialno enačbo rešite za $\theta(t)$ z začetnimi pogoji: $\theta(0) = 0$, $\dot{\theta}(0) = 2\pi\, f$, kjer je $f$ frekvenca vrtenja iz podatkov naloge. Kot rezultat podajte vrednost funkcije $\theta(t_0)$ v obliki števila s plavajočo vejico na $n$ mest natančno. Simbole zamenjajte z vrednostmi iz podatkov naloge ($m$ je masa iz podatkov naloge!). (Namig: začetne pogoje podate z argumentom ``ics={...}`` v funkciji ``sp.dsolve()``, desno stran rešitve dobite z atributom ``.rhs``).

Število poskusov: 3, enote:  rad

```python
resitev = sp.dsolve(
    enacba_gibanja,
    theta,
    ics={
        theta.subs(t_s, 0): 0,
        sp.diff(theta, t_s).subs(t_s, 0): 2 * sp.pi * f_s
    }
)

odgovor18 = resitev.rhs.subs({
    m_s: m,
    R_s: R,
    f_s: f,
    t_s: t_0
}).evalf(n)
```
