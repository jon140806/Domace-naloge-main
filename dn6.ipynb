### 8. vprašanje

Definirajte funkcijo `slovar_iz_vrstice(datoteka, i)`, ki kot argumenta sprejme pot do tekstovne datoteke `datoteka` ter indeks vrstice `i`in po vzoru prejšnje naloge razdeli niz. Vrednosti naj shrani v slovar, kjer je ključ pripadajoč niz iz glave, vrednost pa pripadajoč niz iz seznama telesa prebrane vrstice. V kolikor meji (Meja_1 in Meja_2) ne obstajata, jih ne vključi v slovar. Vrne naj slovar. <br>Posredujte rezultat klica funkcije `slovar_iz_vrstice(datoteka='OEP_data.txt', i=b)`.

Število poskusov: 3, enote:  /

```python
def slovar_iz_vrstice(datoteka, i):
    with open(datoteka, encoding='utf-8') as f:
        vrstice = f.readlines()

    glava = vrstice[0].rstrip('\n').split('\t')
    vrednosti = vrstice[i].rstrip('\n').split('\t')

    return {
        kljuc: vrednost
        for kljuc, vrednost in zip(glava, vrednosti)
        if kljuc not in ('Meja_1', 'Meja_2') or vrednost.strip()
    }

odgovor8 = slovar_iz_vrstice(datoteka='OEP_data.txt', i=b)
```

```python
poslji(odgovor8, id=3055075151912, st=8)
```

### 9. vprašanje

Definirajte funkcijo `integral_iz_datoteke(datoteka, i)`, ki z uporabo `slovar_iz_vrstice` pripravi slovar (`s`) `i`-te vrstice v datoteki. Z uporabo  `sym.symbols` naj vrednost ključa `Simbol` pretvori v simbol (`x = sym.symbols(s['Simbol'])`), s `sym.sympify` pa vrednost ključa `Izraz` pretvotri v simbolni izraz (`izraz = sym.sympify(s['Izraz'])`). Izračuna naj nedoločen integral izraza po definiranemu simbolu in vrne rezultat v obliki niza.<br>Posredujte rezultat klica funkcije `integral_iz_datoteke('OEP_data.txt', a)`.

Število poskusov: 3, enote:  /

```python
import sympy as sym

def integral_iz_datoteke(datoteka, i):
    s = slovar_iz_vrstice(datoteka, i)
    x = sym.symbols(s['Simbol'])
    izraz = sym.sympify(s['Izraz'])
    integral = sym.integrate(izraz, x)
    return str(integral)

odgovor9 = integral_iz_datoteke('OEP_data.txt', a)
```

### 10. vprašanje

Funkcijo `integral_iz_datoteke`, nadgradite, da bo podpirala računanje določenih integralov.<br>Če slovar nima ključev `Meja_1` in `Meja_2`, delovanje funkcije ostane nespremenjeno. V kolikor ključa obstajata ju pretvorite v simbolna izraza (`sp_meja = sym.sympify(s['Meja_1'])` ...) in z uporabo SymPy izračunajte določeni integral. Rezultat vrnite v obliki niza.<br>Posredujte rezultat klica `integral_iz_datoteke('OEP_data.txt', b)`.

Število poskusov: 3, enote:  /

```python
def integral_iz_datoteke(datoteka, i):
    s = slovar_iz_vrstice(datoteka, i)
    x = sym.symbols(s['Simbol'])
    izraz = sym.sympify(s['Izraz'])

    if 'Meja_1' in s and 'Meja_2' in s:
        sp_meja = sym.sympify(s['Meja_1'])
        zg_meja = sym.sympify(s['Meja_2'])
        integral = sym.integrate(izraz, (x, sp_meja, zg_meja))
    else:
        integral = sym.integrate(izraz, x)

    return str(integral)

odgovor10 = integral_iz_datoteke('OEP_data.txt', b)
```
